# 🎛️ All scikit‑image Filters – Cascadable (cell version)

Each filter has its own cell with sliders.  
Filters are applied to the *current* image in sequence – build your own pipeline.  
Use the **Reset** button (top) to revert to the original.

## 1. Imports & Global State

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, filters, restoration, morphology, exposure, color, feature, measure
from skimage.util import img_as_float, img_as_ubyte
import ipywidgets as widgets
from IPython.display import display, clear_output
import warnings
warnings.filterwarnings('ignore')

# Global state
original_image = None
current_image = None
def normalize_image(img):
    if img.dtype == np.uint8:
        img = img_as_float(img)
    if img.ndim == 3 and img.shape[2] == 4:
        img = img[:,:,:3]
    return img

def display_images(original, filtered, title="Current"):
    fig, (ax0, ax1) = plt.subplots(1, 2, figsize=(12, 5))
    ax0.imshow(original, cmap='gray' if original.ndim==2 else None)
    ax0.set_title('Original (reference)')
    ax0.axis('off')
    ax1.imshow(filtered, cmap='gray' if filtered.ndim==2 else None)
    ax1.set_title(title)
    ax1.axis('off')
    plt.tight_layout()
    plt.show()
```

## 4. All Filters (one cell each)

Run any cell to show its sliders and Apply button.  
The filter will be applied to the current image.

```python
def save_img(filename="filtered_image.png"):
    """Saves the current modified image to a file."""
    global current_image
    if current_image is None:
        print("No image available to save.")
        return

    try:
        # Save image using skimage or matplotlib
        plt.imsave(filename, current_image, cmap="gray" if current_image.ndim == 2 else None)
        print(f"Image saved successfully as '{filename}'!")
    except Exception as e:
        print(f"Error saving image: {e}")

def make_filter_cell(name, func, param_defs, is_grayscale=False, is_threshold=False):
    # This function is called inside each cell to set up the UI.
    sliders = {}
    for p in param_defs:
        if 'choices' in p:
            w = widgets.Dropdown(options=p['choices'], value=p['default'], description=p['desc'])
        elif all(isinstance(v, int) for v in [p['min'], p['max'], p['step']]):
            w = widgets.IntSlider(min=p['min'], max=p['max'], step=p['step'], value=p['default'], description=p['desc'])
        else:
            w = widgets.FloatSlider(min=p['min'], max=p['max'], step=p['step'], value=p['default'], description=p['desc'])
        sliders[p['name']] = w
        display(w)

    out = widgets.Output()

    # --- SAVE IMAGE UI CONTROLS ---
    filename_input = widgets.Text(
        value="output_image.png",
        description="Filename:",
        placeholder="Enter filename..."
    )
    save_btn = widgets.Button(
        description="Save Current Image",
        button_style="success",
        icon="save"
    )

    def on_save_clicked(b):
        with out:
            save_img(filename_input.value)

    save_btn.on_click(on_save_clicked)

    # Display Save controls alongside your Output widget
    display(widgets.HBox([filename_input, save_btn]))
    display(out)

    # --- FILTER APPLICATION ---
    clear_output(wait=True)
    kwargs = {k: v.value for k, v in sliders.items()}
    img = current_image

    if img is not None:
        if is_grayscale and img.ndim == 3:
            img = color.rgb2gray(img)
        if is_threshold:
            thresh = func(img, **kwargs)
            binary = img > thresh
            apply_filter_and_update(lambda x, **kw: binary, **{})
        else:
            if 'channel_axis' in func.__code__.co_varnames:
                kwargs['channel_axis'] = -1 if img.ndim == 3 else None
            apply_filter_and_update(func, **kwargs)


def apply_filter_and_update(filter_func, **kwargs):
    global current_image
    if current_image is None:
        print("Load an image first.")
        return
    try:
        filtered = filter_func(current_image, **kwargs)
        current_image = filtered
        clear_output(wait=True)
        display_images(original_image, current_image, title="After filter")
    except Exception as e:
        print(f"Error: {e}")


def prepare_image_for_filter(img, is_grayscale=False):
    if is_grayscale and img.ndim == 3:
        return color.rgb2gray(img)
    return img
```

### frangi (vesselness)

```python
# ========================================================
# Inline frangi filter setup (Interactive & Corrected)
# ========================================================
import ipywidgets as widgets
from skimage import filters, color
# from IPython.display import clear_output, display

# ----- Parameter definitions -----
param_defs = [
    {'name':'scale_range', 'min':1, 'max':30, 'step':1, 'default':10, 'desc':'Scale range (max)'},
    {'name':'scale_step',  'min':1, 'max':5,  'step':1, 'default':2,  'desc':'Step'},
    {'name':'alpha',       'min':0.1, 'max':1.0, 'step':0.05, 'default':0.5, 'desc':'Alpha'},
    {'name':'beta',        'min':0.1, 'max':1.0, 'step':0.05, 'default':0.5, 'desc':'Beta'},
    {'name':'gamma',       'min':0.01,'max':0.5, 'step':0.01, 'default':0.1, 'desc':'Gamma'}
]

# ----- Create sliders -----
sliders = {}
for p in param_defs:
    if 'choices' in p:
        w = widgets.Dropdown(options=p['choices'], value=p['default'], description=p['desc'])
    elif all(isinstance(v, int) for v in [p['min'], p['max'], p['step']]):
        w = widgets.IntSlider(min=p['min'], max=p['max'], step=p['step'],
                              value=p['default'], description=p['desc'], continuous_update=False)
    else:
        w = widgets.FloatSlider(min=p['min'], max=p['max'], step=p['step'],
                                value=p['default'], description=p['desc'], continuous_update=False)
    sliders[p['name']] = w

# ----- Create UI Buttons & Outputs -----
apply_btn = widgets.Button(description="Apply Filter", button_style="primary", icon="check")
save_btn = widgets.Button(description="Save Current Image", button_style="success", icon="save")
filename_input = widgets.Text(value="output_image.png", description="Filename:")

ui_controls = widgets.VBox(list(sliders.values()) + [apply_btn, widgets.HBox([filename_input, save_btn])])
out_image = widgets.Output() # Output area for the image display

# Global variable to hold the latest filtered result
filtered_image = None

# ----- Callbacks -----
def on_apply_clicked(b):
    global filtered_image, current_image, original_image

    with out_image:
        clear_output(wait=True)

        # Fallback check assuming original_image exists globally
        if 'original_image' not in globals() or original_image is None:
            print("Error: 'original_image' is not defined. Load an image first.")
            return

        # Always apply filter to the ORIGINAL image to avoid cumulative filtering
        img = original_image.copy()

        # Grayscale conversion
        if img.ndim == 3:
            img = color.rgb2gray(img)

        # Build kwargs
        kwargs = {k: v.value for k, v in sliders.items()}
        kwargs['scale_range'] = (1, kwargs['scale_range'])

        # Add channel_axis if the function supports it
        if 'channel_axis' in filters.frangi.__code__.co_varnames:
            kwargs['channel_axis'] = None

        try:
            print("Applying Frangi filter... please wait.")
            filtered = filters.frangi(img, **kwargs)
            filtered_image = filtered
            current_image = filtered # Update current_image for subsequent pipeline steps

            clear_output(wait=True)
            # Make sure display_images is defined in your notebook
            display_images(original_image, filtered_image, title="After Frangi filter")
        except Exception as e:
            print(f"Error applying filter: {e}")

def on_save_clicked(b):
    with out_image:
        if filtered_image is not None:
            # Make sure save_img is defined in your notebook
            save_img(filename_input.value, filtered_image)
            print(f"Saved as {filename_input.value}")
        else:
            print("No filtered image to save yet! Click 'Apply Filter' first.")

# ----- Bind callbacks and display -----
apply_btn.on_click(on_apply_clicked)
save_btn.on_click(on_save_clicked)

display(ui_controls)
display(out_image)

# Optionally trigger the filter once on load:
on_apply_clicked(None)
```

### wiener (restoration)

```python
def wiener_wrapper(img, noise=0.1):
    if img.ndim == 3:
        img = color.rgb2gray(img)
    return restoration.wiener(img, np.ones((3,3))/9, noise=noise)

make_filter_cell('wiener', wiener_wrapper,
    [{'name':'noise', 'min':0.0, 'max':0.5, 'step':0.01, 'default':0.1, 'desc':'Noise power'}], is_grayscale=True)
```

### apply_hysteresis_threshold

```python
def hysteresis_wrapper(img, low=0.1, high=0.2):
    if img.ndim == 3:
        img = color.rgb2gray(img)
    return filters.apply_hysteresis_threshold(img, low, high)

make_filter_cell('apply_hysteresis_threshold', hysteresis_wrapper,
    [{'name':'low', 'min':0.0, 'max':0.5, 'step':0.01, 'default':0.1, 'desc':'Low threshold'},
     {'name':'high', 'min':0.0, 'max':1.0, 'step':0.01, 'default':0.2, 'desc':'High threshold'}], is_grayscale=True)
```

### threshold_local

```python
def local_threshold_wrapper(img, block_size=35, method='gaussian', offset=0.0):
    if img.ndim == 3:
        img = color.rgb2gray(img)
    return filters.threshold_local(img, block_size, method=method, offset=offset) > 0

make_filter_cell('threshold_local', local_threshold_wrapper,
    [{'name':'block_size', 'min':5, 'max':101, 'step':2, 'default':35, 'desc':'Block size'},
     {'name':'method', 'choices':['gaussian', 'mean', 'median'], 'default':'gaussian', 'desc':'Method'},
     {'name':'offset', 'min':-0.5, 'max':0.5, 'step':0.01, 'default':0.0, 'desc':'Offset'}], is_grayscale=True)
```

### threshold_niblack

```python
def niblack_wrapper(img, window_size=15, k=0.2):
    if img.ndim == 3:
        img = color.rgb2gray(img)
    return filters.threshold_niblack(img, window_size=window_size, k=k) > 0

make_filter_cell('threshold_niblack', niblack_wrapper,
    [{'name':'window_size', 'min':5, 'max':51, 'step':2, 'default':15, 'desc':'Window size'},
     {'name':'k', 'min':0.0, 'max':1.0, 'step':0.05, 'default':0.2, 'desc':'k'}], is_grayscale=True)
```

### threshold_sauvola

```python
def sauvola_wrapper(img, window_size=15, k=0.2, r=None):
    if img.ndim == 3:
        img = color.rgb2gray(img)
    return filters.threshold_sauvola(img, window_size=window_size, k=k, r=r) > 0

make_filter_cell('threshold_sauvola', sauvola_wrapper,
    [{'name':'window_size', 'min':5, 'max':51, 'step':2, 'default':15, 'desc':'Window size'},
     {'name':'k', 'min':0.0, 'max':1.0, 'step':0.05, 'default':0.2, 'desc':'k'},
     {'name':'r', 'min':0.1, 'max':1.0, 'step':0.05, 'default':0.5, 'desc':'r (dynamic range)'}], is_grayscale=True)
```

### threshold_multiotsu (produces label image)

```python
def multiotsu_wrapper(img, classes=3):
    if img.ndim == 3:
        img = color.rgb2gray(img)
    thresh = filters.threshold_multiotsu(img, classes=classes)
    regions = np.digitize(img, bins=thresh)
    return regions / (classes-1)  # scale to [0,1] for display

make_filter_cell('threshold_multiotsu', multiotsu_wrapper,
    [{'name':'classes', 'min':2, 'max':10, 'step':1, 'default':3, 'desc':'Number of classes'}], is_grayscale=True)
```

### rank_order (normalized ranks)

```python
def rank_order_wrapper(img):
    if img.ndim == 3:
        img = color.rgb2gray(img)
    return filters.rank_order(img)[0] / np.max(img)

make_filter_cell('rank_order', rank_order_wrapper, [], is_grayscale=True)
```

### gabor (real part)

```python
def gabor_wrapper(img, frequency=0.1, theta=0, sigma_x=1, sigma_y=1, offset=0):
    if img.ndim == 3:
        img = color.rgb2gray(img)
    real, imag = filters.gabor(img, frequency=frequency, theta=theta, sigma_x=sigma_x, sigma_y=sigma_y, offset=offset)
    return real

make_filter_cell('gabor (real)', gabor_wrapper,
    [{'name':'frequency', 'min':0.01, 'max':0.5, 'step':0.01, 'default':0.1, 'desc':'Frequency'},
     {'name':'theta', 'min':0, 'max':np.pi, 'step':0.1, 'default':0, 'desc':'Theta (rad)'},
     {'name':'sigma_x', 'min':0.5, 'max':5.0, 'step':0.5, 'default':1.0, 'desc':'Sigma x'},
     {'name':'sigma_y', 'min':0.5, 'max':5.0, 'step':0.5, 'default':1.0, 'desc':'Sigma y'},
     {'name':'offset', 'min':0, 'max':2*np.pi, 'step':0.1, 'default':0, 'desc':'Offset'}], is_grayscale=True)
```
